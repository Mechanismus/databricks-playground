```sql
-- My CSV files landed as tables in the wrong schema: hive_metastore.default instead of hive_metastore.bronze_salesforce; now I'll be moving them

CREATE TABLE hive_metastore.bronze_salesforce.account
AS SELECT * FROM hive_metastore.default.account;

CREATE TABLE hive_metastore.bronze_salesforce.contact
AS SELECT * FROM hive_metastore.default.contact;

CREATE TABLE hive_metastore.bronze_salesforce.task
AS SELECT * FROM hive_metastore.default.task;

CREATE TABLE hive_metastore.bronze_salesforce.case
AS SELECT * FROM hive_metastore.default.case;

-- Drop the old ones (optional)
DROP TABLE hive_metastore.default.account;
DROP TABLE hive_metastore.default.contact;
DROP TABLE hive_metastore.default.task;
DROP TABLE hive_metastore.default.case;

DROP TABLE hive_metastore.default.sales_demo 
```
